```julia
using Revise
using MajoranaPropagation
using PauliPropagation
```

$\mu(v_1) = \gamma_1 \gamma_1^\prime \gamma_3 \gamma_6^\prime$

$\mu(v_2) = \gamma_1 \gamma_2^\prime \gamma_3 \gamma_4^\prime$

```julia
n_fermions = 8 

v1 = [1, 2, 5, 12]
msum1 = MajoranaSum(Float64, n_fermions, v1)

v2 = [1, 4, 5, 8]
msum2 = MajoranaSum(Float64, n_fermions, v2)
```

```julia
msum = msum1 * msum2
```

Spinful case

$\mu(v_3) = \gamma_{1,\uparrow} \gamma_{2,\uparrow} \gamma_{3,\downarrow}^\prime \gamma_{4,\uparrow}^\prime$

```julia
n_spinful_sites = 4
gammas_up = [1, 3, 8]
gammas_down = [6]

msum = MajoranaSum(Float64, n_spinful_sites, gammas_up, gammas_down)
```
